### AI-Driven FINANCIAL CRISIS EARLY WARNING SYSTEM (KENYA)

Think of these system like a smoke detector for the economy.

It doesn't stop the fire it warns early so that action can be taken

**1. Business context**

Kenya is exposed to macro-economic shocks driven by:
- Rising inflation
- Exchange rate volatility(KES/USD)
- Fluctuating Foreign reserves

These shocks:
- Reduce Purchasing power
- Increase Debt servicing cost
- Pressure banks and public finances
- Can escalate into a Financial or balance of payment crisis

Currently warning signs exists in data, but:
- They are mentioned individually
- Insights are reactive
- Decision makers often respond after stress has escalated

**2. Business Problem**

Financial crises are detected too late.

By the time:
- Inflation spikes,
- The shilling Depreciates sharply,
- Reserves fall below critical levels

The cost of intervention becomes very high.

There is no intergrated,data-driven early warning system that continuosly asses macro-economic stress and translates it into actionable risk signals.

**3. Business Objective**

The objective of these project is to:
**Design an AI-based early warning system that detects rising financial strees in the Kenyan economy and signals elevated crisis risks before a full-scale crisis occurs**

**4. Intended Business Value**

The system is intended to:
- Provide early alerts(Months not days)
- Support pro-active policy decisions
- Reduce the economic costs of late interventions
- Improve confidence in macro-economic monitoring

These is a **decision-support tool**, not ana automated policy engine

**5. Stake Holders and Users**

**Primary Stake Holders**
- Central Bank Of Kenya
- National Treasury
- Financial sector regulators

**Secondary Stake Holders**
- Commercial Banks
- Institutional Investors
- Economic Resaerchers

These users require:
- Simple signals
- Clear explanations
- Low false-alarm noise

**6. Defination of a Financial Crisis(Business View)**

For these project, a financial crisis is defined as:

**A period of sustained macroeconomic instability characterized by sharp currency depreciation, accelerating inflation, and weakening foreign reserves adequacy, which threatens Financial and economic stability**

These defination aligns with:
- Balance of payment stress
- Currency crises
- Inflation-driven instability


**7. Key Business Indicators**

The system focuses on 3 core indicators:

**1. Inflation(CPI)**
- Measures internal price pressure
- Indicates loss of purchasing power

**2. Exchange Rate(KES/USD)**
- Reflects external confidence and capital flow
- Signals currency stress

**3. Foreign Reserves**
- Represents the economy defensive capability
- Determines ability to stabilize the currency

Individually important--**collectively Powerful.**  


**8. Early Warning Concept**

The system does not attemt to:
- Predict exact crisis dates
- Replace economic judgement

Instead, it:
- Detects abnormal changes
- Indentifies dangerous combinations
- Assigns a crisis risk level

**Business-friendly outputs**
- Low risk
- Moderate risk
- High risk
- Critical risk

**Decisions Supported**

When risks rises the systems enables:

**Central Bank Action**
- FX market intervention planning
- Interest rates adjustments
- Reserve management strategies

**Government actions**
- Fiscal tightening 
- Import controls or subsidies
- Early engagements with development partners

**Business Constraints and Risks**

**Constraints**
- Limited historical crisis periods
- Data frequency and revision delays 
- Structural economic changes over time

**Risks**
- False Positives causing unnecessary concerns
- Political sensitivity of crisis signals

To mitigate this:
- Outputs are advisory
- Results are interpretable
- Alerts ar tiered, not binary

**11.Business Success Criteria**

The project is successful if:
- Rising risk is detected before visible market stress
- Alerts are understandable by non-technical users
- The system detects consistency across historical stress periods
- False alarms are minimized

**12. Business Framing Statement**

**This project developes an AI-driven financial crisis early warning system that integrates inflation, exchange rate, and foreign reserve data to assess macroeconomic stress and provide timely, actionable risk signals for economic decision-makers in Kenya.**

### Computing annual FX depreciation (%)

**What we are computing(concept first)**

$$
\text{Depreciation}_t
=
\frac{FX_t - FX_{t-1}}{FX_{t-1}} \times 100
$$

interpretation:
- Positive % → KES depreciated

- Negative % → KES appreciated

- This is the true FX “return”

### Create lagged exchange rate ###

```python
fx['fx_lag'] = fx['kes_per_usd'].shift(1)
fx.head()
```

**The result:** The value that was originally at Index 0 moves to Index 1. This means that for any given row, the fx_lag column now contains the exchange rate from the previous period (the year before).

### Compute Depreciation rate

```python
fx['fx_depreciation_pct'] = (
    (fx['kes_per_usd'] - fx['fx_lag']) / fx['fx_lag']
) * 100
```

Merging them on year, then compute some useful metrics.

**The plan:**

- Merge the two dataframes on year using a left join on FX so we keep all FX years.

- Compute annual FX depreciation (%) if not done properly yet.

- Compute reserve coverage metric, e.g., how much a year’s FX depreciation would require from reserves.

```python
# Assuming your FX dataframe is 'fx' and reserves dataframe is 'reserves'

# 1️⃣ Merging on year
merged = fx.merge(df_reserves, on='year', how='left')

# 2️⃣ Computing FX depreciation % if needed
merged['fx_depreciation_pct'] = merged['kes_per_usd'].pct_change() * 100

# 3️⃣ Computing a simple coverage metric
# For example, how much of reserves would be needed to prevent full depreciation in USD
merged['reserves_vs_fx_depreciation'] = merged['reserves_usd'] / merged['kes_per_usd']

# 4️⃣ Inspect
print(merged.head())
print(merged.tail())
```

```python
# Simple coverage: how many years of depreciation (%) could reserves absorb
merged['coverage_years'] = merged['reserves_usd'] / (merged['kes_per_usd'] * 1e7)  # adjusting the  scale
```

```python
# Filtering out rows where fx_depreciation_pct is NaN
plot_data = merged.dropna(subset=['fx_depreciation_pct', 'reserves_usd'])

# Creating the figure and axis
fig, ax1 = plt.subplots(figsize=(12,6))

# Plotting FX depreciation (%) on left y-axis
ax1.plot(plot_data['year'], plot_data['fx_depreciation_pct'], color='red', marker='o', label='FX Depreciation (%)')
ax1.set_xlabel('Year')
ax1.set_ylabel('FX Depreciation (%)', color='red')
ax1.tick_params(axis='y', labelcolor='red')

# Creating a second y-axis for reserves
ax2 = ax1.twinx()
ax2.plot(plot_data['year'], plot_data['reserves_usd'], color='blue', marker='x', label='Reserves (USD)')
ax2.set_ylabel('Reserves (USD)', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')

# Adding a title and grid
plt.title('KES FX Depreciation vs Reserves Over Time')
ax1.grid(True)

# Show legend
fig.tight_layout()
fig.legend(loc='upper left', bbox_to_anchor=(0.1,0.9))

plt.show()
```

```python
# Identifying extremes
max_dep = plot_data['fx_depreciation_pct'].idxmax()  # 2023
min_dep = plot_data['fx_depreciation_pct'].idxmin()  # 2024

# Plotting setup
fig, ax1 = plt.subplots(figsize=(12,6))

# FX Depreciation %
ax1.plot(plot_data['year'], plot_data['fx_depreciation_pct'], color='red', marker='o', label='FX Depreciation (%)')
ax1.set_xlabel('Year')
ax1.set_ylabel('FX Depreciation (%)', color='red')
ax1.tick_params(axis='y', labelcolor='red')

# Highlighting extremes
ax1.scatter(plot_data.loc[max_dep, 'year'], plot_data.loc[max_dep, 'fx_depreciation_pct'],
            color='darkred', s=100, zorder=5, label='Max Depreciation')
ax1.scatter(plot_data.loc[min_dep, 'year'], plot_data.loc[min_dep, 'fx_depreciation_pct'],
            color='green', s=100, zorder=5, label='Appreciation')

# Reserves
ax2 = ax1.twinx()
ax2.plot(plot_data['year'], plot_data['reserves_usd'], color='blue', marker='x', label='Reserves (USD)')
ax2.set_ylabel('Reserves (USD)', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')

# Titles and grid
plt.title('KES FX Depreciation vs Reserves Over Time with Extremes Highlighted')
ax1.grid(True)

# Legend
fig.tight_layout()
fig.legend(loc='upper left', bbox_to_anchor=(0.1,0.9))

plt.show()
```
